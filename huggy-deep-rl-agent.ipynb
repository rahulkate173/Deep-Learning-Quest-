```python
%%bash
mkdir -p /kaggle/working/trained-envs-executables/linux
wget -q "https://github.com/huggingface/Huggy/raw/main/Huggy.zip" -O /kaggle/working/trained-envs-executables/linux/Huggy.zip
unzip -q -o /kaggle/working/trained-envs-executables/linux/Huggy.zip -d /kaggle/working/trained-envs-executables/linux/
chmod -R 755 /kaggle/working/trained-envs-executables/linux/Huggy
```

```python
# verification of ml agents lib 
!/kaggle/working/mlagent_env/bin/mlagents-learn --help
```

```python
%%writefile /kaggle/working/ml-agents/config/ppo/Huggy.yaml
behaviors:
  Huggy:
    trainer_type: ppo
    hyperparameters:
      batch_size: 2048 # num of env
      buffer_size: 20480 #
      learning_rate: 0.0003 # lr as gradient step
      beta: 0.005 # the
      epsilon: 0.2
      lambd: 0.95
      num_epoch: 3 # steps 
      learning_rate_schedule: linear # lr schedular
    network_settings:
      normalize:  # normalize 
      hidden_units: 512
      num_layers: 3
      vis_encode_type: simple
    reward_signals:
      extrinsic:
        gamma: 0.995
        strength: 1.0
    checkpoint_interval: 200000 # checkpoint at every 
    keep_checkpoints: 15
    max_steps: 2000000
    time_horizon: 1000
    summary_freq: 50000
```

```python
# running/ training huggy 
%cd /kaggle/working/ml-agents
!/kaggle/working/mlagent_env/bin/mlagents-learn ./config/ppo/Huggy.yaml \
    --env=/kaggle/working/trained-envs-executables/linux/Huggy/Huggy \
    --run-id="Huggy" \
    --no-graphics
```
