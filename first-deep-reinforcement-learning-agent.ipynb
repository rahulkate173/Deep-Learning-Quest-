The Gymnasium library provides two things:


* An interface that allows you to create RL environments.
* A collection of environments (gym-control, atari, box2D...).


````python
import gymnasium as gym

# First, we create our environment called LunarLander-v2
env = gym.make("LunarLander-v3")

# Then we reset this environment to initial state 
observation, info = env.reset() # 

for _ in range(20):
  # Take a random action
  action = env.action_space.sample() # taken and random actions 
  print("Action taken:", action)

  # Do this action in the environment and get
  # next_state, reward, terminated, truncated and info
  observation, reward, terminated, truncated, info = env.step(action) # for that action get reward,obs,terminated,truncated,info

  # If the game is terminated (in our case we land, crashed) or truncated (timeout)
  if terminated or truncated:
      # Reset the environment
      print("Environment is reset")
      observation, info = env.reset()

env.close()
````

### LunarLander environment 🌛 and understand how it works

````python
# We create our environment with gym.make("<name_of_the_environment>")
env = gym.make("LunarLander-v3")
env.reset()
print("_____OBSERVATION SPACE_____ \n")
print("Observation Space Shape", env.observation_space.shape)
print("Sample observation", env.observation_space.sample()) # Get a random observation
````

We see with `Observation Space Shape (8,)` that the observation is a vector of size 8, where each value contains different information about the lander:
- Horizontal pad coordinate (x)
- Vertical pad coordinate (y)
- Horizontal speed (x)
- Vertical speed (y)
- Angle
- Angular speed
- If the left leg contact point has touched the land (boolean)
- If the right leg contact point has touched the land (boolean)


````python
print("\n _____ACTION SPACE_____ \n")
print("Action Space Shape", env.action_space.n)
print("Action Space Sample", env.action_space.sample()) # Take a random action
````

The action space (the set of possible actions the agent can take) is discrete with 4 actions available 🎮:

- Action 0: Do nothing,
- Action 1: Fire left orientation engine,
- Action 2: Fire the main engine,
- Action 3: Fire right orientation engine.

Reward function (the function that will give a reward at each timestep) 💰:

After every step a reward is granted. The total reward of an episode is the **sum of the rewards for all the steps within that episode**.

For each step, the reward:

- Is increased/decreased the closer/further the lander is to the landing pad.
-  Is increased/decreased the slower/faster the lander is moving.
- Is decreased the more the lander is tilted (angle not horizontal).
- Is increased by 10 points for each leg that is in contact with the ground.
- Is decreased by 0.03 points each frame a side engine is firing.
- Is decreased by 0.3 points each frame the main engine is firing.

The episode receive an **additional reward of -100 or +100 points for crashing or landing safely respectively.**

An episode is **considered a solution if it scores at least 200 points.**

## Create the Model 🤖
- We have studied our environment and we understood the problem: **being able to land the Lunar Lander to the Landing Pad correctly by controlling left, right and main orientation engine**. Now let's build the algorithm we're going to use to solve this Problem 🚀.

- To do so, we're going to use our first Deep RL library, [Stable Baselines3 (SB3)](https://stable-baselines3.readthedocs.io/en/master/).

- SB3 is a set of **reliable implementations of reinforcement learning algorithms in PyTorch**.

---

💡 A good habit when using a new library is to dive first on the documentation: https://stable-baselines3.readthedocs.io/en/master/ and then try some tutorials.

----

Stable-Baselines3 is easy to set up:

1️⃣ You **create your environment** (in our case it was done above)

2️⃣ You define the **model you want to use and instantiate this model** `model = PPO("MlpPolicy")`

3️⃣ You **train the agent** with `model.learn` and define the number of training timesteps

```
# Create environment
env = gym.make('LunarLander-v2')

# Instantiate the agent
model = PPO('MlpPolicy', env, verbose=1)
# Train the agent
model.learn(total_timesteps=int(2e5))
```



````python
# We added some parameters to accelerate the training
model = PPO(
    policy = 'MlpPolicy',
    env = env,
    n_steps = 1024,
    batch_size = 64,
    n_epochs = 4,
    gamma = 0.999,
    gae_lambda = 0.98,
    ent_coef = 0.01,
    verbose=1)
````

## Train the PPO agent 🏃

````python
# Train it for 1,000,000 timesteps
model.learn(total_timesteps=1000000)
# Save the model
model_name = "ppo-LunarLander-v3"
model.save(model_name)
````

The tabular output from `model.learn()` is printed by Stable-Baselines3 every time an on-policy rollout buffer fills up and an update cycle finishes. Each table is split into three main sections: **Rollout**, **Time**, and **Train**.

---

### 1. `rollout/` (Environment Interaction)

These metrics evaluate the agent's actual gameplay performance over the most recent 100 episodes:

* **`ep_rew_mean` (Mean Episode Reward):** The single most important indicator of learning.
* In your run, it starts at **-190** (crashing almost immediately) and steadily climbs to **+81.1** by iteration 25.
* In LunarLander, a score of **+200** or higher is considered "solved." Rising steadily into the positive range means the agent has learned to stabilize and is landing upright more often rather than nose-diving.


* **`ep_len_mean` (Mean Episode Length):** The average number of steps the lander survived before the episode terminated (either by landing, crashing, or timeout).
* Your run went from **~94 steps** at iteration 1 to **880 steps** at iteration 25.
* Longer episodes indicate the agent learned to fire its thrusters to hover and decelerate rather than hitting the ground in free fall.



---

### 2. `time/` (Hardware & Training Speed)

* **`fps` (Frames Per Second):** The throughput of environment transitions processed per second. It begins around **3,600 FPS** and levels out around **1,030–1,050 FPS** as episodes grow longer and vector env handling balances out.
* **`iterations`:** The number of complete collection-and-optimization loops executed. Each iteration collects `n_steps * n_envs` = $1024 \times 16 = 16,384$ timesteps.
* **`time_elapsed`:** Cumulative wall-clock training time in seconds (395s $\approx$ 6.5 minutes at iteration 25).
* **`total_timesteps`:** Total interactions with the environment so far ($409,600$ at iteration 25 out of the target $1,000,000$).

---

### 3. `train/` (Optimization & PPO Losses)

These metrics detail how the neural network's weights are updating during the 4 epochs (`n_epochs=4`) across mini-batches (`batch_size=64`):

* **`explained_variance`:** Measures how well the Critic (value function) predicts the expected future returns.
* Formula: $1 - \frac{\text{Var}(y - \hat{y})}{\text{Var}(y)}$.
* Early on, it was near **0** (or negative), meaning the value network had no idea what returns to expect.
* By iteration 24–25, it hits **~0.91 – 0.95**, meaning the Critic can predict the cumulative return from a state with over 90% accuracy.


* **`value_loss`:** The mean squared error between predicted state values and actual discounted returns. It dropped dramatically from **$5.27 \times 10^3$** down to **~39–97**, showing the Critic network converged nicely.
* **`policy_gradient_loss`:** The clipped surrogate objective of PPO that trains the Actor. Stable values around $-0.001$ to $-0.007$ indicate steady policy improvement without erratic shifts.
* **`approx_kl`:** An approximation of the Kullback-Leibler (KL) divergence between the old policy and the updated policy.
* Staying consistently between **0.004 and 0.010** confirms that PPO's clipping mechanism (`clip_range=0.2`) is successfully preventing destabilizing, overly large policy updates.


* **`clip_fraction`:** The fraction of policy updates where the probability ratio exceeded $1 \pm \text{clip\_range}$. Values around **3% to 4%** are typical and healthy.
* **`entropy_loss`:** The negative policy entropy ($-\mathcal{H}$).
* It shifts from **-1.38** toward **-1.12**. Lower absolute entropy means the policy is transitioning from random exploration toward confident, deterministic landing maneuvers.


* **`loss`:** The total combined objective minimized by PyTorch: $\mathcal{L}_{\text{total}} = \mathcal{L}_{\text{policy}} + c_1 \mathcal{L}_{\text{value}} - c_2 \mathcal{H}$.
* **`learning_rate`:** Fixed at **0.0003** ($3 \times 10^{-4}$), the default Adam optimizer learning rate.
* **`n_updates`:** Total gradient descent passes completed ($96$ updates by iteration 25, which equals $24 \times 4\text{ epochs}$).

- Stable-Baselines3 provides a method to do that: `evaluate_policy`.
- To fill that part you need to [check the documentation](https://stable-baselines3.readthedocs.io/en/master/guide/examples.html#basic-usage-training-saving-loading)

````python
import gymnasium as gym
import imageio
from huggingface_hub import HfApi

# Create evaluation environment for rendering
eval_env = gym.make("LunarLander-v3", render_mode="rgb_array")

frames = []
obs, _ = eval_env.reset()
done = False

# Record one full episode
while not done:
    frame = eval_env.render()
    frames.append(frame)
    action, _states = model.predict(obs, deterministic=True)
    obs, reward, terminated, truncated, info = eval_env.step(action)
    done = terminated or truncated

eval_env.close()

# Save frames as mp4 using imageio (fps=30)
video_path = "replay.mp4"
imageio.mimsave(video_path, frames, fps=30)
print(f"Replay saved: {len(frames)} frames recorded.")
````
